```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Year"] = df["Date"].dt.year
```

```python
df["Month"] = df["Date"].dt.month
```

```python
df["Month_Name"] = df["Date"].dt.month_name()
```

```python
df["Day"] = df["Date"].dt.day
```

```python
df["Day_Name"] = df["Date"].dt.day_name()
```

```python
df["Day_of_Week"] = df["Date"].dt.dayofweek
```

```python
df[["Date", "Year", "Month", "Month_Name", "Day", "Day_Name", "Day_of_Week"]].head()
```

```python
df["Revenue_Per_Passenger"] = df["Revenue"] / df["Passengers"]
```

```python
df["Delay_Category"] = pd.cut(
    df["Delay_Minutes"],
    bins=[-1, 0, 15, 30, float("inf")],
    labels=["On Time", "Slight Delay", "Moderate Delay", "Heavy Delay"]
)
```

```python
df["Distance_Category"] = pd.cut(
    df["Distance_KM"],
    bins=[0, 300, 700, 1500, float("inf")],
    labels=["Short", "Medium", "Long", "Very Long"]
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
top_trains = df.groupby("Train_Name")["Passengers"].sum().sort_values(ascending=False).head(10)
```

```python
plt.figure(figsize=(10, 6))
plt.barh(top_trains.index, top_trains.values)
plt.xlabel("Total Passengers")
plt.ylabel("Train Name")
plt.title("Top 10 Trains by Total Passengers")
plt.show()
```

```python
top_revenue = df.groupby("Train_Name")["Revenue"].sum().sort_values(ascending=False).head(10)
```

```python
plt.figure(figsize=(10, 6))
plt.barh(top_revenue.index, top_revenue.values)
plt.xlabel("Total Revenue")
plt.ylabel("Train Name")
plt.title("Top 10 Trains by Total Revenue")
plt.gca().invert_yaxis()
plt.show()
```
